```
import sys
from pathlib import Path
import numpy as np
import pandas as pd

BASE_DIR = Path("..").resolve() if Path("..").resolve().name == "CricketIQ 2.0" else Path(".").resolve()
sys.path.insert(0, str(BASE_DIR))

from src.data_loader import load_data, CURRENT_TEAMS
from src.squad_data import get_squad_data

matches, deliveries = load_data(BASE_DIR)
squads_2026 = get_squad_data(BASE_DIR)

print("🔍 Auditing Missing Values in Matches Dataset:")
display(pd.DataFrame({
    'Column': matches.columns,
    'Null Count': matches.isnull().sum(),
    'Null Percentage (%)': (matches.isnull().sum() / len(matches) * 100).round(2)
}))
```

### Squad Validation: 2025/2026 Mega-Auction Roster Verification

```
rcb = squads_2026['Royal Challengers Bengaluru']['squad']
gt = squads_2026['Gujarat Titans']['squad']
kkr = squads_2026['Kolkata Knight Riders']['squad']

checks = [
    ("Mohammed Siraj removed from RCB?", 'Mohammed Siraj' not in rcb),
    ("Mohammed Siraj verified in Gujarat Titans?", 'Mohammed Siraj' in gt),
    ("Venkatesh Iyer verified in RCB?", 'VR Iyer' in rcb),
    ("Jacob Duffy verified in RCB?", 'Jacob Duffy' in rcb),
    ("Phil Salt verified in RCB?", 'PD Salt' in rcb),
    ("Faf du Plessis removed from RCB?", 'F du Plessis' not in rcb),
    ("Glenn Maxwell removed from RCB?", 'GJ Maxwell' not in rcb),
    ("Karn Sharma removed from RCB?", 'KV Sharma' not in rcb),
]

audit_df = pd.DataFrame(checks, columns=["Audit Rule", "Pass Status"])
display(audit_df)

assert 'Mohammed Siraj' not in rcb, "Siraj should not be in RCB!"
assert 'Mohammed Siraj' in gt, "Siraj should be in GT!"
assert 'VR Iyer' in rcb, "Venkatesh Iyer should be in RCB!"
print("\n✅ ALL SQUAD INTEGRITY AUDITS PASSED 100%!")
```
